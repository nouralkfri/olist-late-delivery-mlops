```python
import pandas as pd
from sklearn.model_selection import train_test_split

ml_table = pd.read_parquet('ml_table_notebook2.parquet')
ml_table.shape
```

```python
train_val, test = train_test_split(
    ml_table, test_size=0.2, random_state=42, stratify=ml_table['is_late']
)
train, val = train_test_split(
    train_val, test_size=0.2, random_state=42, stratify=train_val['is_late']
)
print("train:", train.shape, "val:", val.shape, "test:", test.shape)
```

```python
train.to_parquet('train.parquet', index=False)
val.to_parquet('val.parquet', index=False)
test.to_parquet('test.parquet', index=False)
print("saved!")
```
