```python
import pandas as pd

ml_table = pd.read_parquet('ml_table_notebook1.parquet')
ml_table.shape
```

```python
ml_table[['order_status', 'order_delivered_customer_date', 'order_estimated_delivery_date']].head(10)
```

```python
ml_table['order_delivered_customer_date'] = pd.to_datetime(ml_table['order_delivered_customer_date'])
ml_table['order_estimated_delivery_date'] = pd.to_datetime(ml_table['order_estimated_delivery_date'])

ml_table['is_late'] = (ml_table['order_delivered_customer_date'] > ml_table['order_estimated_delivery_date']).astype(int)

ml_table[['order_id', 'order_delivered_customer_date', 'order_estimated_delivery_date', 'is_late']].head(10)
```

```python
late_example = ml_table[ml_table['is_late'] == 1][['order_delivered_customer_date', 'order_estimated_delivery_date', 'is_late']].head(3)
print(late_example)

ontime_example = ml_table[ml_table['is_late'] == 0][['order_delivered_customer_date', 'order_estimated_delivery_date', 'is_late']].head(3)
print(ontime_example)
```

```python
print("عدد الطلبات بدون تاريخ تسليم فعلي:", ml_table['order_delivered_customer_date'].isna().sum())
```

```python
print("قبل الحذف:", ml_table.shape)

ml_table = ml_table[ml_table['order_delivered_customer_date'].notna()].copy()

print("بعد الحذف:", ml_table.shape)
```
